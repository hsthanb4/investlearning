【课程定位与投决会核心心智模型】
::: callout-note
### 📌 投决会核心心智模型：制度性错杀是超额 Alpha 的温床，清算价值是绝对安全边际底座
二级市场存在大量的非基本面制度性摩擦。当大型企业集团剥离非核心子公司进行独立分拆上市（Spinoff）时，由于机构投资者受限于基金投资范围契约（如大盘基金无法持有分拆出的小微盘股票、指数基金无脑剔除非成分股等），引发非理性的机构无差别集中踩踏抛售，形成极端低估的‘孤儿股’。本杰明·格雷厄姆（Benjamin Graham）的 Net-Net 营运资本模型（Net Current Asset Value, NCAV）构成了困境资产定价的物理防线：完全剔除所有厂房设备与商誉，仅以现金 100%、应收账款 75%、存货 50% 扣除全部负债，即便公司明天立即清算，股东分得的净现金仍显著大于当前市值。
:::

- **[企业分拆的戴维斯双击动能]**：分拆消除了母公司的多元化折价（Conglomerate Discount），子公司管理层通常获配极为丰厚且直接挂钩业绩的股权期权，在抛压消化后的 1~3 年内爆发强劲的 ROIC 扩张。
- **[困境反转的赔率与仓位铁律]**：困境重组标的具有 5~10 倍的高赔率，但伴随着破产清算归零的极端尾部风险；投决会组合管理严格限定单笔仓位 $\le 2\%$，依靠凸性期权属性实现组合非对称收益。
- **[债转股与资产负债表出清]**：企业破产重整中的‘债转股（Debt-to-Equity Swap）’将刚性付息债务永久转化为权益资本，瞬间修复流动性比率并卸下财务费用重担。

- **[要点 1：非经济性抛压的确定性机会]**：机构基金经理卖出并非因为子公司基本面恶化，而是受制于合规投资契约的机械性执行，创造了极度罕见的流动性折价。
- **[要点 2：管理层激励相容的第一性原理]**：在母公司内部，子公司高管只是领薪打工人；独立分拆后，高管直接持有大量行权价低廉的期权，资本配置效率与盈利动力发生根本性跃迁。
- **[要点 3：抛压消化的时间窗口]**：通常分拆后的前 1~3 个月为机构抛压集中出清期，随后进入价值发现的主升浪。

锚点指引：下图展示了格雷厄姆净营运流动资产（Net-Net / NCAV）清算安全边际漏斗。

```mermaid
flowchart LR
    A["流动资产原值"] --> B["折算净变现现金: 100% 现金 + 75% 应收账款 + 50% 存货"]
    B --> C["刚性扣除 100% 全部负债 (短期借款 + 长期负债 + 应付账款)"]
    C --> D["= 净营运安全清算价值 (Net-Net NCAV)"]
    D --> E["投决会准入标准: 市值 Market Cap < 2/3 × NCAV"]
```

- **[要点 1：固定资产估值归零法]**：厂房、机器设备、土地、无形资产与商誉在清算模型中一律计为 0 元，杜绝一切资产虚高泡沫。
- **[要点 2：存货与应收账款的保守折价]**：存货打五折处置、应收账款打七五折计提坏账，提供最极限的防灾安全垫。
- **[要点 3：买入即锁定义务清算暴利]**：当市值低于净清算价值的三分之二时，相当于花 60 块钱买入 100 块钱现金，即便企业经营完全停滞，也是严格正期望投资。

【数学与估值模型】
本杰明·格雷厄姆净营运资产清算价值公式（NCAV / Net-Net）：
$$
\text{NCAV} = 1.00 \times \text{Cash} + 0.75 \times \text{Receivables} + 0.50 \times \text{Inventory} - 1.00 \times \text{Total Liabilities}
$$
投决会绝对安全边际买入准则：
$$
\text{Market Cap} \le \frac{2}{3} \times \text{NCAV} \iff \text{Margin of Safety} \ge 33.3\%
$$
变量定义域与参数约束：
- $\text{Cash}, \text{Receivables}, \text{Inventory}, \text{Total Liabilities} \ge 0$。
- $\text{NCAV} \in (-\infty, +\infty)$：只有当 $\text{NCAV} > 0$ 且远大于市值时才构成合格安全边际。

困境反转凯利仓位约束方程：
$$
f^* = \frac{p \cdot b - (1 - p)}{b}, \quad f_{\text{distressed}} \le \min\left(f^*, \; 0.02\right)
$$
变量定义域：$p$ 为重组成功率（通常仅 $20\% \sim 30\%$），$b$ 为反转赔率（通常 $5 \sim 10$），单票头寸严格封顶在总资产 $2\%$。

::: callout-warning
### ⚠️ 投决会风控红线与认知陷阱防范
1. **价值陷阱与现金流失血（Value Trap & Cash Burn）**：若企业主业严重亏损，经营性现金流持续为负，即便账面满足 Net-Net，其净现金也会在 1~2 年内被管理层薪酬与亏损彻底烧光，必须动态监控现金消耗率（Burn Rate）。
2. **大股东掏空与破产清算黑天鹅**：部分民营暴雷企业存在严重的关联方资金占用与隐性表外负债担保，必须经过严格法务与实地现场穿透审计。
3. **仓位无纪律放大**：严禁因“赔率极高”而在困境反转股票上重仓超配，违规者扣除季度绩效。
:::
