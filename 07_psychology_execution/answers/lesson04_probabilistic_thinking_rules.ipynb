::: callout-note
### 📌 投决会核心心智模型：放弃单次确定性预测，拥抱大数定律离散抽样
专业量化机构与业余散户的本质分野，在于对“不确定性”的认知底座。业余交易者穷极一生追求“100% 预测下一根 K 线走势的神迹”，本质是在物理不可预测的非平稳随机过程中寻找确定性幻觉；而专业投资机构的投决会完全接受“单次交易具有不可约减的随机性”，将每一笔入场视为大数定律（Law of Large Numbers）下的独立伯努利抽样，核心工作仅聚焦于建立并机械执行“正数学期望决策树（Positive Expectancy Decision Tree）”。
:::

- **[单次随机性 vs 总体确定性]**：在单次抛硬币中，没有任何人能断言正反面；但在 10,000 次抛硬币中，正面频率必然收敛于 50%。交易者无需预知下一秒价格走向，亦能凭借大数定律实现复利持续增长。
- **[数学期望准入门槛]**：开仓信号必须通过期望效用准入方程 $E = p \cdot b - (1 - p) \ge E_{\min}$，只有预期收益率显著覆盖流动性摩擦与方差风险溢价时，系统才批准下单。
- **[规则代码化与防篡改]**：主观交易的致命缺陷在于“临场随意性”与“情绪妥协”。唯一能够击败人性弱点的方式，是将概率规则编译为计算机可执行代码，彻底物理剥离人类在开平仓临界的自由裁量权。

- **[要点 2：仓位下注与凯利约束]**：正期望仅代表具备入场资格，下注比例必须经过半凯利（Half-Kelly）约束，防止过度自信引发非线性波动率拖累。
- **[要点 3：全生命周期机械化托管]**：一旦信号通过决策树，开仓、止损与止盈参数全部交由撮合引擎托管，拒绝人工二次干预。

- **[要点 2：方差随交易笔数的相对压缩]**：累计收益均值的标准误以 $1/\sqrt{N}$ 速度衰减，当交易笔数突破 300 笔后，策略真实数学期望占主导，运气方差被有效稀释。
- **[要点 3：复利增长的第一性原理]**：只要单笔期望 $E > 0$ 且单笔风险严格受控，长期复利路径在时间轴上必然呈现不可逆的向上漂移。

【数学与估值模型】
单笔离散交易数学期望效用比率方程：
$$
E = p \cdot b - (1 - p) = p (b + 1) - 1
$$
变量定义域与参数约束：
- $p \in (0, 1)$：信号回测或前瞻评估历史胜率（Win Probability）。
- $b = \frac{\overline{\text{Win}}}{\overline{\text{Loss}}} > 0$：策略预期盈亏比（Payoff Ratio）。
- $E \in (-1, +\infty)$：以单位风险 $R$（Risk Unit）计价的单笔数学期望比率。
- 决策准入准则：$\mathbb{I}_{\text{Trade}} = \mathbb{I}(E \ge E_{\min})$，其中通常设定 $E_{\min} \in [0.15, 0.30]$。

连续时间几何复合增长率与半凯利下注公式：
$$
g(f) = \mu f - \frac{1}{2} \sigma^2 f^2, \quad f^* = \frac{p \cdot b - (1 - p)}{b}, \quad f_{\text{safe}} = \frac{1}{2} f^*
$$
变量定义域：$f \in [0, 1]$ 为下注资金占总净资产比例，$\mu$ 为收益均值，$\sigma$ 为收益率标准差。

::: callout-warning
### ⚠️ 投决会风控红线与认知陷阱防范
1. **寻求百分百胜率的破产死穴**：试图通过指标堆叠过滤掉所有亏损单，必然导致自由度耗尽（过拟合），并在实盘中遭遇未知黑天鹅单边爆仓。
2. **赌徒谬误（Gambler's Fallacy）**：在经历连续 4 次亏损后，误以为‘下一次盈利的概率必然更高’而翻倍加仓。独立抽样中各次结果互不影响，必须坚守固定风险比例。
3. **沉没成本与自我证明心魔**：在市场证明逻辑破产后，为维护自尊心而与市场对赌，违反投决会第一铁律：认错必须果断，资本永不陪葬。
:::

【Q1 深度参考答案】
- **[统计学自由度耗尽与过拟合陷阱]**：金融市场信噪比极低（通常只有 1%~5%）。追求 100% 胜率必须在历史数据中施加无数重复杂的过滤条件，将参数量推高至逼近数据点数量，耗尽系统统计自由度。此类策略在样本外（实盘）面对未见噪声时将瞬间坍塌。
- **[极端左尾风险的隐蔽暴露]**：在真实市场中，唯一能勉强在短期维持 99% 胜率的策略是深度虚值期权无保底裸卖方（Short Deep OTM Options）或死扛不平仓。这种策略在 99% 的平稳日子里赚取微小权利金，但在 1% 的黑天鹅突发时刻遭遇无限下行亏损，一次性击穿整个资本池。
- **[认知资源错配与自尊心反噬]**：追求 100% 确定性使交易员无法接受哪怕一次合规止损，将微小回撤视为人格侮辱，进而在错误发生时启动死扛加仓防卫机制，走向必然破产。

【Q2 深度参考答案】
- **[独立同分布抽样的长期均值收敛]**：大数定律指出，只要策略单笔数学期望 $E = p \cdot b - (1 - p) > 0$，随着交易笔数 $N \to \infty$，实际样本平均收益率以概率 1 收敛于理论期望值 $\mu$。
- **[单笔随机性与总体确定性的对立统一]**：在单笔交易的微观尺度上，价格波动受纳秒级未知订单流扰动，单次盈亏完全是随机的；但只要数以百计地执行正期望规则，随机性方差在时间序列中互相抵消，组合净值必然呈现平滑向上的宏观确定性。
- **[摆脱情绪焦虑的哲学解脱]**：一旦深谙这一原理，交易员便不再纠结于某一次交易是否被止损，止损只是“买下一张具有正期望的彩票所支付的印花税”，从而彻底解除了因单笔浮亏带来的精神内耗。

【Q3 深度参考答案】
- **[数据感知层、策略逻辑层与柜台执行层的物理隔离]**：
  1. **数据与信号生成层**：仅负责根据 Point-in-Time 市场行情计算特征与期望得分，生成无状态抽象信号对象。
  2. **风控网关层（硬约束哨兵）**：运行独立风控守护进程，对单笔风险预算、当日累计亏损、杠杆率上限进行不可逾越的规则校验，一旦违规直接丢弃信号。
  3. **柜台撮合与订单状态机层**：采用 C++ 或纯代码硬编码 FSM，自动生成 OCO 括号委托并送往交易所，交易员终端仅具备只读监控（Read-Only）权限。
- **[权限收敛与审计日志不可篡改]**：任何规则的修改必须经过投决会离线代码评审（Code Review）与双人公钥多重签名合规审计，在物理架构层面杜绝任何盘中凭借一时冲动下移止损或临时加仓的可能性。


````python
from typing import Dict, Any


def should_take_trade(expected_payoff_ratio: float, win_probability: float, min_expectancy_ratio: float = 0.20) -> bool:
    """
    基于概率决策树判断交易是否值得执行:
    期望效用得分 = p * b - (1 - p)
    若期望效用 >= min_expectancy_ratio，则批准执行，否则放弃
    
    :param expected_payoff_ratio: 预期盈亏比 b (例如 2.5 表示预期平均盈利是止损额的 2.5 倍)
    :param win_probability: 预期胜率 p (0.0 ~ 1.0)
    :param min_expectancy_ratio: 最低期望效用门槛 (默认 0.20 R)
    :return: True 批准执行，False 拒绝放弃
    """
    score = win_probability * expected_payoff_ratio - (1.0 - win_probability)
    return score >= min_expectancy_ratio


def calculate_kelly_fraction(expected_payoff_ratio: float, win_probability: float, half_kelly: bool = True) -> float:
    """
    计算凯利最优资金仓位下注比例:
    f* = (p * b - (1 - p)) / b
    
    :param expected_payoff_ratio: 盈亏比 b
    :param win_probability: 胜率 p
    :param half_kelly: 是否采用稳健半凯利 (0.5 * f*)
    :return: 下注资金占比 (0.0 ~ 1.0)
    """
    if expected_payoff_ratio <= 0.0:
        return 0.0
    f_star = (win_probability * expected_payoff_ratio - (1.0 - win_probability)) / expected_payoff_ratio
    if f_star <= 0.0:
        return 0.0
    f_res = 0.5 * f_star if half_kelly else f_star
    return round(min(f_res, 1.0), 4)


if __name__ == "__main__":
    # 基础决策树断言
    assert should_take_trade(2.5, 0.40) == True # 0.4 * 2.5 - 0.6 = 1.0 - 0.6 = 0.4 >= 0.20
    assert should_take_trade(1.5, 0.40) == False # 0.4 * 1.5 - 0.6 = 0.6 - 0.6 = 0.0 < 0.20

    # 凯利公式辅助计算测试
    # p = 0.5, b = 2.0 -> f* = (0.5*2 - 0.5)/2 = 0.5/2 = 0.25 -> half kelly = 0.1250
    k_half = calculate_kelly_fraction(2.0, 0.5, half_kelly=True)
    assert k_half == 0.1250

    print("Lesson 04 (Probabilistic Thinking) Answer Verified!")
````
