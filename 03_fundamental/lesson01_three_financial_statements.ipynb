### 投决会财务第一性原理：会计复式记账法下的跨期财富守恒定律
- **[商业本质公理]**：市值是企业内在价值在二级市场的报价投影。无论市场情绪如何狂躁，企业的最终价值必然锚定其创造自由现金流的长期能力。三张财务报表是透视管理层资本配置效率的绝对事实底册。
- **[跨期动态勾稽守恒]**：
  - 资产负债表反映某一时点的静态存量存底；
  - 利润表反映跨期经营成果的权责发生制增量；
  - 现金流量表穿透应计粉饰，反映真实的收付实现制现金活水。
- **[三大核心勾稽纽带]**：
  - 现金项闭环：期末货币资金 = 期初货币资金 + 净现金流变动（CFO + CFI + CFF）；
  - 留存收益纽带：期末留存收益 = 期初留存收益 + 归母净利润 - 当期派发现金分红；
  - 营运资本调节：间接法经营现金流 CFO = 净利润 + 非现金折旧摊销 - 净营运资本增量（ΔNWC）。
:::

- **[现代五因子杜邦分析 (DuPont 5-Factor) 严格数学分解]**：
  - 净资产收益率（ROE）通过乘积恒等式解构为五个独立反映商业驱动力的正交分量：

$$
\text{ROE} = \left(\frac{\text{Net Income}}{\text{EBT}}\right) \times \left(\frac{\text{EBT}}{\text{EBIT}}\right) \times \left(\frac{\text{EBIT}}{\text{Revenue}}\right) \times \left(\frac{\text{Revenue}}{\text{Total Assets}}\right) \times \left(\frac{\text{Total Assets}}{\text{Equity}}\right)
$$

- **[变量定义域与五因子经济学内涵]**：
  - $\frac{\text{Net Income}}{\text{EBT}} \in (0, 1]$：税收负担率（Tax Burden）。反映企业税收筹划、研发加计扣除与高新技术税收优惠壁垒。
  - $\frac{\text{EBT}}{\text{EBIT}} \in (0, 1]$：利息负担率（Interest Burden）。反映有息负债利息支出对息税前经营利润的侵蚀程度。
  - $\frac{\text{EBIT}}{\text{Revenue}} \in (0, 1)$：营业利润率（Operating Margin）。核心商业护城河与上下游定价权的关键映射。
  - $\frac{\text{Revenue}}{\text{Total Assets}} \in \mathbb{R}^+$：资产周转率（Asset Turnover）。衡量管理层对固定资产、存货与供应链的运营周转效率。
  - $\frac{\text{Total Assets}}{\text{Equity}} \ge 1.0$：权益乘数 / 财务杠杆（Financial Leverage）。放大收益的同时等比例放大破产清算风险。

- **[引导句锚点 1]**：企业三张财务报表的底层跨期动态勾稽平衡闭环拓扑如下所示：

```mermaid
flowchart LR
    BalanceSheet["资产负债表 (静态资产存底)"] --> CashTie["货币资金项闭环: 期末Cash = 期初Cash + ΔCash"]
    BalanceSheet --> RETie["留存收益项闭环: 期末RE = 期初RE + 净利润 - 分红"]
    IncomeStmt["利润表 (权责发生制)"] --> RETie
    CashFlowStmt["现金流量表 (收付实现制)"] --> CashTie
    IncomeStmt --> CFOTie["间接法调节: CFO ≈ 净利润 + 折旧摊销 - ΔNWC"]
    CFOTie --> CashFlowStmt
```

- **[架构要点拆解 1]**：
  - **[1. 资产负债表绝对平衡]**：资产总计严格恒等于负债总计与股东权益之和，任何一笔虚假资产均需负债或利润配平。
  - **[2. 净利润向权益结转]**：经营赚取的真金白银扣减分红后全数注入资产负债表所有者权益，构成复利增长基石。
  - **[3. 现金流量表真实试金石]**：通过间接法扣减存货和应收账款占用（ΔNWC），揭示企业不依赖外部融资的自主造血能力。

- **[引导句锚点 2]**：现代杜邦五因子分解树状架构与商业驱动力解构如下所示：

```mermaid
flowchart TD
    ROE["ROE 净资产收益率 (股东财富复合增速)"] --> NetMargin["净利润率驱动引擎"]
    ROE --> Efficiency["资产周转率驱动引擎 (Costco 模式)"]
    ROE --> Leverage["财务杠杆驱动引擎 (高危金融杠杆)"]
    NetMargin --> Tax["税收负担: NI / EBT"]
    NetMargin --> Interest["利息负担: EBT / EBIT"]
    NetMargin --> EBITMargin["经营利润率: EBIT / Rev (茅台壁垒)"]
    Efficiency --> AssetTO["总资产周转率: Rev / Assets"]
    Leverage --> EqMult["权益乘数: Assets / Equity"]
```

- **[架构要点拆解 2]**：
  - **[1. 护城河定价权]**：贵州茅台等顶级企业凭借极高的 EBIT 利润率（>60%）单引擎驱动卓越 ROE，完全无需依赖财务杠杆。
  - **[2. 极致周转效率]**：Costco 等零售企业虽然利润率微薄，但通过高达 3.5 次以上的总资产周转率，同样能够创造 20%+ 的高质量 ROE。
  - **[3. 脆弱杠杆陷阱]**：传统房企与爆雷民企通过高达 6~8 倍的权益乘数虚构高 ROE 假象，经济下行期利息支出直接将利润清零。
