【理论精讲】
三张表的灵魂勾稽关系：
1. 资产负债表：资产 = 负债 + 所有者权益（恒等式）
2. 留存收益变动：期末留存收益 = 期初留存收益 + 净利润 - 现金分红
3. 现金流量表：期末货币资金 = 期初货币资金 + 经营活动净现金流 + 投资活动净现金流 + 筹资活动净现金流

杜邦五因子分解模型（Five-Factor DuPont）：
    ROE = (净利润 / 税前利润) * (税前利润 / EBIT) * (EBIT / 营业收入) * (营业收入 / 总资产) * (总资产 / 净资产)
        = 税收负担 * 利息负担 * 经营利润率 * 资产周转率 * 权益乘数
