【理论精讲】
1. 拍卖市场理论（Auction Market Theory, AMT）与价格成交量分布：
   传统技术分析只关注时间维度的成交量（Volume on Time Axis），无法展示不同价格水平上的筹码沉淀。
   成交量分布（Volume Profile, VP）将成交量投射到垂直价格轴（Volume on Price Axis），揭示买卖双方在不同价格区间的博弈筹码密集度。

2. 成交量分布三大核心锚点：
   - POC (Point of Control，控制点):
     整个统计区间内成交量最大的单一价格档位。代表多空双方换手最充分、最认可的微观公允均衡价值中枢（Fair Value Center）。
   - 价值区（Value Area, VA，基于正态分布 1 倍标准差 68.2% 经验规则）：
     以 POC 为核心，向两侧拓展涵盖总成交量 70% 的核心价格区间。
   - VAH (Value Area High，价值区上沿): 价值区的价格上限。
   - VAL (Value Area Low，价值区下沿): 价值区的价格下限。
   市场 80% 的时间在价值区内部进行均值回归博弈；一旦价格脱离价值区且伴随放量，将开启单边趋势突破。

3. 成交量加权平均价 (VWAP) 与标准差偏离通道：
   - 典型价格 Typical Price: TP_t = (High_t + Low_t + Close_t) / 3
   - 累积 VWAP:
     VWAP_t = sum_{i=1}^t (TP_i * Volume_i) / sum_{i=1}^t (Volume_i)
   - 加权标准差 (Weighted Std):
     σ_t = sqrt[ sum_{i=1}^t Volume_i * (TP_i - VWAP_t)^2 / sum_{i=1}^t Volume_i ]
   - 统计偏离通道:
     Upper_Band_k = VWAP_t + k * σ_t
     Lower_Band_k = VWAP_t - k * σ_t
     偏离度标准化分数 Z-Score: Z_t = (TP_t - VWAP_t) / σ_t。
   - 均值回归套利逻辑：
     在无突发宏观外生冲击的日内常态交易中，当 Z_t > +2.0 (严重高估偏离) 时建立空头偏离对冲；当 Z_t < -2.0 (严重低估超卖) 时建立多头反弹对冲，并在价格回归至 VWAP 均线处止盈平仓。
